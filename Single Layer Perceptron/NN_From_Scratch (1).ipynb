```python
x=df.drop(columns='class')
y=df['class']
X_train,X_test,Y_train,Y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
display(X_train.shape)
display(X_test.shape)
display(Y_train.shape)
display(Y_test.shape)
```

```python
class Perceptron:
  def __init__(self,learning_rate=0.3,epochs=300):
    self.learning_rate=learning_rate
    self.epochs=epochs

  def fit(self,X,y):
    n_samples,n_features=X.shape
    self.weights=np.zeros(n_features)
    self.bias=0
    self.errors=[]
    self.weights_history=[]
    self.bias_history=[]
    for epoch in range(self.epochs):
      errors=0
      for x,target in zip(X,y):
        z=np.dot(x,self.weights)+self.bias
        prediction=self.step(z)
        error=target-prediction
        self.weights=self.weights+self.learning_rate*error*x
        self.bias=self.bias+self.learning_rate*error
        if(error!=0):
          errors+=1
      print(f"Epoch {epoch+1}: Errors = {errors} weights = {self.weights} Bias = {self.bias}")
      self.errors.append(errors)
      self.weights_history.append(self.weights)
      self.bias_history.append(self.bias)
  def step(self,z):
    if(z>=0):
      return 1
    else:
      return 0
  def predict(self,X):
    predictions=[]
    for x in X:
      z=np.dot(self.weights,x)+self.bias
      prediction=self.step(z)
      predictions.append(prediction)
    return np.array(predictions)

# class Perceptron:
#   def __init__(self,learning_rate=0.01,epochs=100):
#     self.learning_rate=learning_rate
#     self.epochs=epochs

#   def fit(self,X,y):
#     n_samples,n_features=X.shape
#     self.weights=np.zeros(n_features)
#     self.bias=0
#     self.errors=[]
#     self.weights_history=[]
#     self.bias_history=[]
#     for epoch in range(self.epochs):
#       errors=0
#       indices=np.random.permutation(n_samples)
#       for idx in indices:
#         x=X[idx]
#         target=y.iloc[idx] if hasattr(y,"iloc") else y[idx]
#         z=np.dot(x,self.weights)+self.bias
#         prediction=self.step(z)
#         error=target-prediction
#         self.weights=self.weights+self.learning_rate*error*x
#         self.bias=self.bias+self.learning_rate*error
#         if(error!=0):
#           errors+=1
#       print(f"Epoch {epoch+1}: Errors = {errors} weights = {self.weights} Bias = {self.bias}")
#       self.errors.append(errors)
#       self.weights_history.append(self.weights)
#       self.bias_history.append(self.bias)
#   def step(self,z):
#     if(z>=0):
#       return 1
#     else:
#       return 0
#   def predict(self,X):
#     predictions=[]
#     for x in X:
#       z=np.dot(self.weights,x)+self.bias
#       prediction=self.step(z)
#       predictions.append(prediction)
#     return np.array(predictions)
```

```python
model=Perceptron(learning_rate=0.3,epochs=300)
model.fit(X_train,Y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy=accuracy_score(Y_test,y_pred)
print(accuracy)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

y_pred = model.predict(X_test)

print("Accuracy :", accuracy_score(Y_test, y_pred))
print("Precision:", precision_score(Y_test, y_pred))
print("Recall   :", recall_score(Y_test, y_pred))
print("F1 Score :", f1_score(Y_test, y_pred))

print("\nConfusion Matrix")
print(confusion_matrix(Y_test, y_pred))
```

```python
learning_rates = [0.001, 0.01, 0.1]
for lr in learning_rates:
    model_lr = Perceptron(
        learning_rate=lr,
        epochs=300
    )
    model_lr.fit(X_train, Y_train)
    plt.plot(model_lr.errors, label=f"LR={lr}")
    Y_pred=model_lr.predict(X_test)
    acc=accuracy_score(Y_test,Y_pred)
    print(f"Accuracy with lr {lr} : {acc}")
    print("Weights : ",model_lr.weights)
    print("Bias : ",model_lr.bias)
plt.xlabel("Epoch")
plt.ylabel("Misclassified Samples")
plt.title("Learning Rate Comparison")
plt.legend()
plt.grid(True)
plt.show()
```
